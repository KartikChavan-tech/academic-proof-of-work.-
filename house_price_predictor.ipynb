```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Dataset: House Area (sq. ft.) vs Price (in Lakhs)
data = {
    'Area_sqft': [450, 600, 750, 900, 1100, 1250, 1400, 1600, 1800, 2100],
    'Price_Lakhs': [30, 42, 50, 62, 75, 88, 98, 112, 125, 145]
}
df = pd.DataFrame(data)

# 2. Separate Features (X) and Target (y)
X = df[['Area_sqft']]
y = df['Price_Lakhs']

# 3. Train Linear Regression Model
model = LinearRegression()
model.fit(X, y)

# 4. Model Predictions & Evaluation
y_pred = model.predict(X)
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("--- GDG Model Results ---")
print(f"Slope (Weight): {model.coef_[0]:.4f}")
print(f"Intercept (Bias): {model.intercept_:.4f}")
print(f"R2 Score: {r2:.4f}")

# 5. Plot Graph (Actual Data vs Regression Line)
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', label='Actual House Prices', s=50)
plt.plot(X, y_pred, color='red', linewidth=2, label='Linear Regression Prediction Line')
plt.xlabel('House Size / Area (sq. ft.)')
plt.ylabel('Price (in Lakhs INR)')
plt.title('House Price Prediction Model - GDG Pre-Interview Task')
plt.legend()
plt.grid(True)
plt.savefig('house_price_prediction.png')
plt.show()
```
